```python
#list
a=[10,20,30,40,50,60]
b=[1,2,3]

print("list 1:",a)
print("list 2:",b)
a.append(40)
a.extend([70,80])
b.extend([4,6])
print("updated list :",a)
a.remove(20)
print("After remove:",a)
a.pop(3)
print("After pop:",a)

print("After pop at index 1:",a)
print("Index of 80:",a.index(80))
print("lenght of a:",len(a))
print("lenght of b:",len(b))
print("Elements in list:")
print("Elements i list 1:",a)
print("Elements in list 2:",b)

```
